# 01 NAV & Returns Analysis 📈

This notebook demonstrates how to fetch NAV history, plot prices, and calculate trailing and rolling returns for Indian Mutual Funds.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from mf_analyser.data.cache import get_nav
from mf_analyser.analysis.returns import returns_table, rolling_returns, max_drawdown
from mf_analyser.config import DEFAULT_FUNDS, FUND_NAME_TO_CODE

# Pre-configured styling
pd.options.plotting.backend = "plotly"
```

## 1. Fetch & Plot NAV
Let's select **PPFAS FlexiCap Fund**.

```python
fund_name = "PPFAS FlexiCap Fund"
scheme_code = FUND_NAME_TO_CODE[fund_name]

nav_df = get_nav(scheme_code)
print(f"Loaded {len(nav_df)} records for {fund_name}")

fig = px.line(nav_df, x="date", y="nav", title=f"{fund_name} NAV History")
fig.show()
```

## 2. Trailing Returns
Standard performance metrics (1Y, 3Y, 5Y, etc.)

```python
rt = returns_table(nav_df)
rt
```

## 3. Rolling Returns (3Y CAGR)
Rolling returns give a better picture of the consistency of the fund.

```python
window = 3
rdf = rolling_returns(nav_df, window_years=window)
col = f"rolling_{window}y_cagr_pct"

fig = px.line(rdf, x="date", y=col, title=f"{fund_name} - Rolling {window}Y CAGR (%)")
fig.add_hline(y=rdf[col].mean(), line_dash="dash", annotation_text="Average")
fig.show()
```
