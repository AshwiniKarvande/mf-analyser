# 03 AUM Analysis 🏦

Analyze Assets Under Management (AUM) trends and sub-scheme distributions.

```python
import pandas as pd
import plotly.express as px
from mf_analyser.aum.tracker import scheme_aum_trend, aum_growth_summary, subscheme_aum_split
from mf_analyser.data.cache import get_aum

QUARTER = "Jan-Mar 2024"
FUND_KEYWORD = "PPFAS"
```

## 1. Sub-scheme AUM Split
Mutual funds have Direct/Regular and Growth/IDCW versions. Let's see how much AUM each holds.

```python
quarter_df = get_aum(QUARTER)
split_df = subscheme_aum_split(quarter_df, FUND_KEYWORD)
split_df
```

```python
fig = px.pie(split_df, values="aum_cr", names="scheme_name", 
             title=f"AUM Breakdown for {FUND_KEYWORD} ({QUARTER})")
fig.show()
```

## 2. AUM Trend (Historical)
Note: This relies on mftool having historical AUM data for specified quarters.

```python
# Tracking PPFAS Flexi Cap Fund (Direct Growth specifically)
trend_df = scheme_aum_trend("PPFAS Flexi Cap Fund - Direct Plan - Growth", start_year=2022)
trend_df = aum_growth_summary(trend_df)
trend_df
```

```python
fig = px.bar(trend_df, x="quarter", y="aum_cr", title="AUM Growth Trend (Crores)", text_auto=True)
fig.show()
```
