# 02 Strategy Backtesting 🤖

Compare different investment strategies (SIP, Lump Sum, Value Averaging, Momentum) on historical data.

```python
import pandas as pd
import plotly.graph_objects as go
from mf_analyser.data.cache import get_nav
from mf_analyser.analysis.strategies import (lump_sum, sip, value_averaging, 
                                          momentum_ma, sip_with_stop_loss, compare_strategies)
from mf_analyser.config import FUND_NAME_TO_CODE

fund_name = "Mirae Asset Large Cap Fund"
scheme_code = FUND_NAME_TO_CODE[fund_name]
nav_df = get_nav(scheme_code)

START_DATE = "2019-01-01"
```

## 1. Run Multiple Strategies

```python
results = []

# 1. Lump Sum (1 Lakh)
results.append(lump_sum(nav_df, amount=100000, start_date=START_DATE))

# 2. SIP (5000/mo)
results.append(sip(nav_df, monthly_amount=5000, start_date=START_DATE))

# 3. Value Averaging (Target 5000/mo @ 1% growth)
results.append(value_averaging(nav_df, monthly_target_growth=1.0, start_amount=5000, start_date=START_DATE))

# 4. Momentum (MA50/200)
results.append(momentum_ma(nav_df, amount=100000, fast_window=50, slow_window=200, start_date=START_DATE))

# 5. SIP + Stop-Loss (20% Drop)
results.append(sip_with_stop_loss(nav_df, monthly_amount=5000, stop_loss_pct=20.0, start_date=START_DATE))

comparison_df = compare_strategies(results)
comparison_df
```

## 2. Compare Performance (CAGR %)

```python
import plotly.express as px
fig = px.bar(comparison_df, x="strategy", y="cagr_pct", color="strategy", 
             title=f"Strategy Comparison (CAGR %) - {fund_name}", text_auto=True)
fig.show()
```
